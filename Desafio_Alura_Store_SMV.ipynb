#1. Ingresos



```python
ventas_por_tienda = tiendas.groupby('Tienda')['Precio'].sum().reset_index()
ventas_por_tienda
```

```python
# Función para convertir número en millones
def millones(x,pos):
  return f'${x/1e6:,.1f}M'
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Gráfico con Seaborn
plt.figure(figsize=(10, 6))
ax=sns.barplot(data=ventas_por_tienda, x='Tienda', y='Precio', palette='viridis')

# Valores sobre las barras
for p in ax.patches:
    ax.annotate(millones(p.get_height(),0),
                (p.get_x() + p.get_width() / 2., p.get_height()+18000000),
                ha='center', va='center', fontsize =12)

ax.yaxis.set_major_formatter(millones)

plt.title('Ingresos por Tienda (2020-2023)', fontsize=16)
plt.xlabel('Tienda')
plt.ylabel('Ingresos')

plt.show()
```

```python
plt.pie(ventas_por_tienda['Precio'], labels=ventas_por_tienda['Tienda'], autopct='%1.1f%%')
```

*   La tienda 1 es la que genera mayores ingresos.
*   La tienda 4 es la que genera menores ingresos.
* Todas las tiendas tienen ventas mayores a 1000 millones entre 2020 y marzo de 2023.

## Facturación por Año

Hay que recordar que el año 2023 sólo considera hasta Marzo.

```python
ventas_tienda_ano = tiendas.groupby(['Tienda','Año'])['Precio'].sum().reset_index()
ventas_tienda_ano
```

```python
# Gráfico con Seaborn
plt.figure(figsize=(10, 6))
ax=sns.barplot(data=ventas_tienda_ano, x='Año', y='Precio', hue='Tienda', palette='rocket')

# Valores sobre las barras
for p in ax.patches:
      if p.get_width() > 0:
        ax.annotate(millones(p.get_height(),0),
                    (p.get_x() + p.get_width() / 2., p.get_height()+40000000),
                    ha='center', va='center', fontsize =12, rotation=90)

ax.yaxis.set_major_formatter(millones)

plt.ylim(0, 500000000)
plt.title('Ingresos por Tienda (2020-2023)', fontsize=16)
plt.xlabel('Tienda')
plt.ylabel('Ingresos')


plt.show()
```

```python
tiendas['AñoMes'] = tiendas['Fecha de Compra'].dt.to_period('M')  # Crear periodo año-mes
ventas_mensuales = tiendas.groupby(['AñoMes', 'Tienda'])['Precio'].sum().reset_index()
```

```python
plt.figure(figsize=(16, 8))

ax = sns.lineplot(
    data=ventas_tienda_ano,
    x='Año',
    y='Precio',
    hue='Tienda',
    style='Tienda',
    markers=True,
    dashes=False,
    linewidth=2.5,
    markersize=10
)

plt.title('Evolución Anual de Ingresos por Tienda', fontsize=18, pad=20, weight='bold')
plt.xlabel('Periodo (Año)', fontsize=14, labelpad=15)
plt.ylabel('Ingresos Totales', fontsize=14, labelpad=15)

years = sorted(ventas_tienda_ano['Año'].unique())
plt.xticks(years,fontsize=11)
plt.yticks(fontsize=11)
plt.legend(title='Tienda', fontsize=12, title_fontsize=14)
ax.xaxis.set_major_formatter(plt.FormatStrFormatter('%d'))

plt.show()
```

```python
ventas_mensuales['AñoMes'] = ventas_mensuales['AñoMes'].astype(str)
stores = ventas_mensuales['Tienda'].unique()


fig, axes = plt.subplots(2, 2, figsize=(20, 12))
fig.suptitle('Evolución Mensual de Ingresos por Tienda', fontsize=20, weight='bold', y=1.02)

# Aplanar el array de ejes para iterar
axes = axes.flatten()

# Graficar una tienda por subplot
for i, stores in enumerate(stores[:4]):  # Asegurar que solo ploteamos 4 tiendas
    ax = axes[i]
    datos_tienda = ventas_mensuales[ventas_mensuales['Tienda'] == stores]

    sns.lineplot(
        data=datos_tienda,
        x='AñoMes',
        y='Precio',
        color='steelblue',  # Color uniforme (opcional)
        marker='o',        # Marcadores para puntos
        linewidth=2.5,
        markersize=10,
        ax=ax
    )

    ax.set_title(f'Tienda: {stores}', fontsize=16, pad=15)
    ax.set_xlabel('Periodo (Año-Mes)', fontsize=12, labelpad=10)
    ax.set_ylabel('Ingresos Totales', fontsize=12, labelpad=10)
    ax.tick_params(axis='x', rotation=45, labelsize=10)
    ax.tick_params(axis='y', labelsize=10)

    # Mostrar solo algunos ticks en el eje x para evitar saturación
    n_ticks = min(12, len(datos_tienda['AñoMes'].unique()))  # Máximo 12 ticks
    ax.xaxis.set_major_locator(plt.MaxNLocator(n_ticks))

# Ajustar espaciado entre subplots
plt.tight_layout()
plt.show()
```

# 3. Calificación promedio de la tienda


```python
calificacion_tienda = tiendas.groupby(['Tienda'])['Calificación'].mean().reset_index()
calificacion_tienda
```

```python
# Comparación de Medianas

import scipy.stats as stats
# Test de Kruskal-Wallis
h_stat, p_valor = stats.kruskal(tienda1['Calificación'], tienda2['Calificación'], tienda3['Calificación'], tienda4['Calificación'])

print(f"Estadístico H = {h_stat:.3f}, p-valor = {p_valor:.4f}")

if p_valor < 0.05:
    print("Hay diferencias significativas entre las tiendas (rechazamos H₀).")
else:
    print("No hay diferencias significativas (no rechazamos H₀).")
```

En términos de clasificación, es la tienda 1 la que presenta una menor clasificación promedio. Al comparar la medianas, podemos ver que no hay diferencias significativas entre las calificaciones de las tiendas.

## Calificación Promedio por Año de las tiendas

```python
calificacion_tienda_ano = tiendas.groupby(['Tienda','Año'])['Calificación'].mean().reset_index()
```

```python
calificacion_tienda_ano.pivot(index='Tienda', columns='Año', values='Calificación').round(2)
```

```python
plt.figure(figsize=(6, 4))

heatmap_data = calificacion_tienda_ano.pivot(index='Tienda', columns='Año', values='Calificación')
sns.heatmap(heatmap_data, cmap='rocket', annot=True, fmt='.2f', linewidths=0.5)

plt.title('Mapa de calor de calificación promedio por tienda', fontsize=12)
plt.xlabel('Tienda')
plt.ylabel('Año')
plt.tight_layout()
plt.show()
```

Podemos observar que la tienda 1 a pesar de haber tenido la calificación promedio menor, no es la que tiene la evaluación menor cada año.
